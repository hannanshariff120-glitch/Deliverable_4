#### Task 1: Temperature class

describtion of code/class, focus on how we check for the wanted values

```python
class Temperature:

    def __init__(self, value, unit):
        unit = unit.upper()
        if unit != "C" and unit != "K" and unit != "F":
            raise ValueError("The temperature must be in C, K or F")
        self._unit = unit
        if unit == "K" and value < 0:
            raise ValueError("The temperature can't be bellow absolute zero")
        if unit == "C" and value < -273.15:
            raise ValueError("The temperature can't be bellow absolute zero")
        if unit == "F" and value < -459.67:
            raise ValueError("The temperature can't be bellow absolute zero")
        self._value = value

    @property
    def value(self):
        return self._value

    @property
    def unit(self):
        return self._unit

    def to_celsius(self) -> float:
        if self._unit == "C":
            return self._value
        elif self._unit == "K":
            return self._value - 273.15
        return (self._value - 32) * 5 / 9

    def to_kelvin(self) -> float:
        if self._unit == "K":
            return self._value
        elif self._unit == "C":
            return self._value + 273.15
        return (self._value - 32) * 5 / 9 + 273.15

    def to_fahrenheit(self) -> float:
        if self._unit == "F":
            return self._value
        elif self._unit == "C":
            return self._value * 9 / 5 + 32
        return (self._value - 273.15) * 9 / 5 + 32

    def convert_to(self, target_unit: str):
        if target_unit.upper() == "C":
            c_value = self.to_celsius()
            return Temperature(c_value, target_unit)

        elif target_unit.upper() == "K":
            k_value = self.to_kelvin()
            return Temperature(k_value, target_unit)

        elif target_unit.upper() == "F":
            f_value = self.to_fahrenheit()
            return Temperature(f_value, target_unit)

        raise ValueError("The temperature must be in C, K or F")

    def __repr__(self):
        return f"Temperature({self._value:.1f}, '{self._unit}')"

    def __str__(self):
        return f"{self._value:.2f}°{self._unit}"

    def __lt__(self, other) -> bool:
        if type(other) is not type(self):
            raise NotImplementedError("You must compare two temperatures")
        new_other = other.convert_to(self.unit)
        return new_other.value > self.value

    def __gt__(self, other) -> bool:
        if type(other) is not type(self):
            raise NotImplementedError("You must compare two temperatures")
        new_other = other.convert_to(self.unit)
        return new_other.value < self.value

    def __le__(self, other) -> bool:
        if type(other) is not type(self):
            raise NotImplementedError("You must compare two temperatures")
        new_other = other.convert_to(self.unit)
        return new_other.value >= self.value

    def __ge__(self, other) -> bool:
        if type(other) is not type(self):
            raise NotImplementedError("You must compare two temperatures")
        new_other = other.convert_to(self.unit)
        return new_other.value <= self.value

    def __eq__(self, other) -> bool:
        if type(other) is not type(self):
            raise NotImplementedError("You must compare two temperatures")
        new_other = other.convert_to(self.unit)
        if abs(new_other.value - self.value) <= 10**(-9):
            return True
        return False
```

#### Task 1: Testing the temperature class

Describe how the tests work

```python
import pytest
from Temperature import Temperature


def test_case_insensitive_unit():
    temp = Temperature(25, "c")
    assert temp.value == 25
    assert temp.unit == "C"


@pytest.mark.parametrize("unit, expected_unit", [("c", "C"),
                                                 ("k", "K"),
                                                 ("f", "F"),
                                                 ("C", "C"),
                                                 ("K", "K"),
                                                 ("F", "F")])
def test_case_all_units(unit, expected_unit):
    temp = Temperature(25, unit)
    assert temp.value == 25
    assert temp.unit == expected_unit


def test_invalid_unit_raises_exception():
    with pytest.raises(ValueError, match="The temperature must be in C, K or F"):
        Temperature(25, "p")


def test_read_only_behavior_for_value():
    with pytest.raises(AttributeError):
        temp = Temperature(25, "c")
        temp.value = 30


def test_read_only_behavior_for_unit():
    with pytest.raises(AttributeError):
        temp = Temperature(25, "c")
        temp.unit = "F"


@pytest.mark.parametrize("value, unit", [(-300, "C"),
                                         (-500, "F"),
                                         (-1, "K")])
def test_temperatures_below_absolute_zero(value, unit):
    with pytest.raises(ValueError):
        Temperature(value, unit)


@pytest.mark.parametrize("celsius, fahrenheit, kelvin", [(-273.15, -459.67, 0),
                                                         (-40, -40, 233.15),
                                                         (0, 32, 273.15),
                                                         (37, 98.6, 310.15),
                                                         (100, 212, 373.15)])
def test_convertion_accuracy(celsius, fahrenheit, kelvin):
    d_C = Temperature(celsius, "C")
    d_F = Temperature(fahrenheit, "F")
    d_K = Temperature(kelvin, "K")
    delta = 10**(-9)

    assert abs(d_C.to_kelvin() - d_K.value) <= delta
    assert abs(d_C.to_fahrenheit() - d_F.value) <= delta

    assert abs(d_F.to_celsius() - d_C.value) <= delta
    assert abs(d_F.to_kelvin() - d_K.value) <= delta

    assert abs(d_K.to_celsius() - d_C.value) <= delta
    assert abs(d_K.to_fahrenheit() - d_F.value) <= delta


def test_round_trip_convertion():
    celsius = Temperature(25, "C")
    fahrenheit = celsius.convert_to("F")
    kelvin = fahrenheit.convert_to("K")
    new_celsius = kelvin.convert_to("C")
    delta = 10**(-9)
    assert abs(celsius.value - new_celsius.value) <= delta

def test_cross_unit_comparisons():
    kelvin = Temperature(25, "K")
    celsius = Temperature(25, "C")

    fahrenheit = Temperature(-40, "F")
    celsius_cross = Temperature(-40, "C")

    assert kelvin < celsius
    assert celsius > kelvin
    assert celsius != kelvin
    assert fahrenheit == celsius_cross
    assert fahrenheit <= celsius_cross
    assert fahrenheit >= celsius_cross
    assert celsius_cross >= fahrenheit
    assert celsius_cross <= fahrenheit


def test_formatting_outputs():
    celsius = Temperature(25, "C")
    assert str(celsius) == "25.00°C"
    assert repr(celsius) == "Temperature(25.0, 'C')"
```

#### Task 2: Creating the fraction class

Describe how the class works, focus on how we check the values are correct

```python
class Fraction:

    def __init__(self, numerator: int, denominator: int):
        if type(numerator) is not int:
            raise TypeError("The numerator should be a integer not a any other type")
        self.numerator = numerator
        if type(denominator) is not int:
            raise TypeError("The denominator should be a integer not any other type")
        elif denominator <= 0:
            raise ValueError("The denominator has to be larger than zero")
        self.denominator = denominator

    @property
    def value(self):
        return self.numerator / self.denominator

    def __lt__(self, other):
        if type(other) is not type(self):
            raise TypeError("You must compare to fractions")
        return self.value < other.value

    def __gt__(self, other):
        if type(other) is not type(self):
            raise TypeError("You must compare to fractions")
        return self.value > other.value

    def __eq__(self, other):
        if type(other) is not type(self):
            raise TypeError("You must compare to fractions")
        return self.value == other.value

    def __le__(self, other):
        if type(other) is not type(self):
            raise TypeError("You must compare to fractions")
        return self.value <= other.value

    def __ge__(self, other):
        if type(other) is not type(self):
            raise TypeError("You must compare to fractions")
        return self.value >= other.value

    def __str__(self):
        return f"{self.numerator} / {self.denominator}"

    def __repr__(self):
        return f"Fraction({self.numerator}, {self.denominator})"

    def __add__(self, other):
        if self.denominator == other.denominator:
            new_numerator = self.numerator + other.numerator
            return Fraction(new_numerator, self.denominator)

        new_denominator = other.denominator * self.denominator
        new_other_nom = other.numerator * self.denominator
        new_self_nom = self.numerator * other.denominator
        new_numerator = new_other_nom + new_self_nom

        return Fraction(new_numerator, new_denominator)

    def __sub__(self, other):
        if self.denominator == other.denominator:
            new_numerator = self.numerator - other.numerator
            return Fraction(new_numerator, self.denominator)

        new_denominator = other.denominator * self.denominator
        new_other_nom = other.numerator * self.denominator
        new_self_nom = self.numerator * other.denominator
        new_numerator = new_self_nom - new_other_nom

        return Fraction(new_numerator, new_denominator)
```

#### Task 2: Testing the fraction class

Describe how the tests work

```python
import pytest
from Fractions import Fraction


def test_fraction_comparisons():
    fract1 = Fraction(2, 3)
    fract2 = Fraction(4, 6)
    fract3 = Fraction(1, 2)

    assert fract1 == fract2
    assert fract2 == fract1

    assert fract1 != fract3
    assert fract3 != fract1

    assert fract2 > fract3
    assert fract3 < fract2

    assert fract2 >= fract3
    assert fract3 <= fract2

    assert fract2 >= fract1
    assert fract2 <= fract1


def test_formatting_outputs():
    fract = Fraction(1, 2)
    assert str(fract) == "1 / 2"
    assert repr(fract) == "Fraction(1, 2)"


def test_addition_same_denominator():
    fract1 = Fraction(1, 3)
    fract2 = Fraction(2, 3)

    result = fract1 + fract2
    expected_result = Fraction(3, 3)

    assert result == expected_result


def test_addition_different_denominator():
    fract1 = Fraction(1, 3)
    fract2 = Fraction(2, 4)

    result = fract1 + fract2
    expected_result = Fraction(10, 12)

    assert result == expected_result


def test_subtraction_same_denominator():
    fract1 = Fraction(1, 3)
    fract2 = Fraction(2, 3)

    result = fract1 - fract2
    expected_result = Fraction(-1, 3)

    assert result == expected_result


def test_subtraction_different_denominator():
    fract1 = Fraction(1, 3)
    fract2 = Fraction(2, 4)

    result = fract1 - fract2
    expected_result = Fraction(-2, 12)

    assert result == expected_result


def test_negative_denominator():
    with pytest.raises(ValueError):
        Fraction(1, -3)


def test_float_values():
    with pytest.raises(TypeError, match="The numerator should be a integer not a any other type"):
        Fraction(1.4, 3)
    with pytest.raises(TypeError, match="The denominator should be a integer not any other type"):
        Fraction(1, 3.5)
```
